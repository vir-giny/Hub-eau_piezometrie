# 04_fusion-EDA_stations-chroniques  

4 règles :  
- **Nettoyer chaque source séparément, avant la fusion**: chaque source a son notebook de nettoyage et produit un fichier propre.  
- **Harmoniser les clés de jointure** : Les colonnes qui servent à relier les tables doivent avoir le même nom, le même type et le même format partout.  
- **Définir le « grain » de la table récap avant de coder** : Le grain, c’est ce que représente une ligne.  
    - une ligne par station → table de synthèse (niveau moyen, tendance, nombre de mesures…) ;  
    - une ligne par station × année → pour suivre l’évolution ;  
    - une ligne par station × jour → table détaillée pour l’analyse temporelle.  
Si deux tables n’ont pas le même grain, il faut d’abord agréger la plus fine (groupby), puis joindre.  
- **Contrôler chaque jointure** : Dans 03_fusion, après chaque merge, vérifie   
    - print(len(gauche), len(droite), len(resultat))   # le nombre de lignes a-t-il explosé ou fondu ?
    - pd.merge(..., how="left", validate="one_to_one", indicator=True)  # détecte les doublons et les non-correspondances
    - resultat["_merge"].value_counts()   # combien de lignes sans correspondance ?

NB : Sauvegarde les fichiers intermédiaires en Parquet (df.to_parquet()) plutôt qu’en CSV. Les types sont conservés (les dates restent des dates), et c’est plus léger et plus rapide
