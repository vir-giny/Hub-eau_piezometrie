# 03_analyse-EDA_stations  
 
- **EDA univarié** : moyenne, médiane,écart-type, quartiles, min et max, forme de la distribution, variables catégorielles ? 1 phrases d'observation par graphique
- **EDA bivarié** : graphique suivant le type de deux variables et suivant la "question métier" de la relation entre eux ?  
- **EDA multivariée** : matrice de corrélation (heatmap), graphiques par groupe, carte géolocalisé avec couleurs par indicateurs.  
- **Analyse temporelle** : évolution de la courbe, saisonnalité, tendance, normalisation ?  
- **Analyse inférentielle** : Hypothèse formulée clairement : H0 (« pas de tendance ») contre H1 (« tendance à la baisse »). Test choisi selon les données (Mann-Kendall, pente de Sen, Mann-Whitney / Kruskal-Wallis, Spearman)


### PLAN DU NOTEBOOK

**Objectif** : à partir des données brutes téléchargées par `01_collectAPI_stationss`, contrôler leur qualité et analyser.

- Source : API Hub'Eau « Stations » (https://hubeau.eaufrance.fr/page/api-piezometrie)
- Le projet se fait en quatres notebooks :
    1. `01_collectAPI_stations` et `01_collectAPI_chroniques`: liste des stations de la zone (endpoint `/stations`) et téléchargement des mesures de chaque station (endpoint `/chroniques`);
    2. `02_nettoyage_description_stations.ipynb` et `02_nettoyage_description_chroniques` : contrôle qualité, préparation et analyse des mesures
    3. **`03_analyse-EDA_stations` (ce notebook)** : analyse univarié et bivarié, graphique de visualisation et analyse temporelle, analyse inférentielle.  
    4. `04_fusion-EDA_stations-chroniques.ipynb` : fusion des tables
