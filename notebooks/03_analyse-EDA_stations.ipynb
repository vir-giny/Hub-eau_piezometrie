# 03_analyse-EDA_stations  
 
- **EDA univarié** : moyenne, médiane,écart-type, quartiles, min et max, forme de la distribution, variables catégorielles ? 1 phrases d'observation par graphique
- **EDA bivarié** : graphique suivant le type de deux variables et suivant la "question métier" de la relation entre eux ?  
- **EDA multivariée** : matrice de corrélation (heatmap), graphiques par groupe, carte géolocalisé avec couleurs par indicateurs.  
- **Analyse temporelle** : évolution de la courbe, saisonnalité, tendance, normalisation ?  
- **Analyse inférentielle** : Hypothèse formulée clairement : H0 (« pas de tendance ») contre H1 (« tendance à la baisse »). Test choisi selon les données (Mann-Kendall, pente de Sen, Mann-Whitney / Kruskal-Wallis, Spearman)
